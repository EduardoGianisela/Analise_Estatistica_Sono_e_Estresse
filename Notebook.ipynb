##Análise Exploratória de Dados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Base de dados
df = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")

df.head()
```

```python
#Selecionando 200 linhas aleatórias
df = df.sample(200, random_state=42)
print("Novo tamanho da base:", df.shape)
```

```python
print("Dimensões da base:", df.shape)
df.info()
df.describe()
```

```python
#MÉDIAS, MEDIANAS E MODAS

print("Média Sleep Duration:", round(df["Sleep Duration"].mean(), 2))
print("Mediana Sleep Duration:", round(df["Sleep Duration"].median(), 2))
print("Moda Sleep Duration:", round(df["Sleep Duration"].mode()[0], 2))

print("\nMédia Heart Rate:", round(df["Heart Rate"].mean(), 2))
print("Mediana Heart Rate:", round(df["Heart Rate"].median(), 2))
print("Moda Heart Rate:", round(df["Heart Rate"].mode()[0], 2))

print("\nMédia Daily Steps:", round(df["Daily Steps"].mean(), 2))
print("Mediana Daily Steps:", round(df["Daily Steps"].median(), 2))
print("Moda Daily Steps:", round(df["Daily Steps"].mode()[0], 2))

```

```python
print("Quartis Sleep Duration:")
print(df["Sleep Duration"].quantile([0.25, 0.50, 0.75]))

print("\nQuartis Heart Rate:")
print(df["Heart Rate"].quantile([0.25, 0.50, 0.75]))

print("\nQuartis Daily Steps:")
print(df["Daily Steps"].quantile([0.25, 0.50, 0.75]))
```

```python
plt.hist(df["Age"])
plt.ylabel("Frequência")
plt.show()

plt.hist(df["Sleep Duration"])
plt.title("Histograma - Sleep Duration")
plt.xlabel("Horas de Sono")
plt.ylabel("Frequência")
plt.show()

plt.hist(df["Heart Rate"])
plt.title("Histograma - Frequência Cardíaca")
plt.xlabel("BPM")
plt.ylabel("Frequência")
plt.show()

# Atividade Física
plt.hist(df["Physical Activity Level"])
plt.ylabel("Frequência")
plt.show()
```

```python
corr = df[['Age','Sleep Duration','Heart Rate','Quality of Sleep','Stress Level','Physical Activity Level']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, fmt=".2f", cmap="Blues")
plt.title("Mapa de calor – Correlação entre Variáveis")
plt.show()
```

```python
plt.scatter(df["Sleep Duration"], df["Stress Level"])
plt.xlabel("Horas de Sono")
plt.ylabel("Nível de Estresse")
plt.show()

plt.scatter(df["Stress Level"], df["Heart Rate"])
plt.xlabel("Nível de Estresse")
plt.ylabel("Frequência Cardiaca (bpm)")
plt.show()

plt.scatter(df["Sleep Duration"], df["Heart Rate"])
plt.title("Relação entre Sleep Duration e Heart Rate")
plt.xlabel("Sleep Duration (h)")
plt.ylabel("Heart Rate (bpm)")
plt.show()

plt.scatter(df["Physical Activity Level"], df["Heart Rate"])
plt.xlabel("Atividade Física")
plt.ylabel("Batimentos cardiacos")
plt.show()

plt.scatter(df["Age"], df["Stress Level"])
plt.xlabel("Idade")
plt.ylabel("Nível de Estresse")
plt.show()
```
